## 🔹 Gradient-Based Methods

These methods assume the likelihood function is **differentiable**. Most suitable when gradients are available or can be estimated efficiently (e.g., via automatic differentiation).

### 1. **Gradient Descent (GD)**

- **Use When**: The likelihood is smooth, and the parameter space is not too high-dimensional.
- **Pros**: Simple; widely available.
- **Cons**: Slow convergence; sensitive to learning rate.

### 2. **Stochastic Gradient Descent (SGD)**

- **Use When**: Large datasets; online or mini-batch updates are desirable.
- **Pros**: Scalable; memory-efficient.
- **Cons**: Noisy updates; requires careful tuning.

### 3. **Newton-Raphson**

- **Use When**: The likelihood is twice differentiable and moderate in dimensionality.
- **Pros**: Quadratic convergence near optimum.
- **Cons**: Requires Hessian; expensive in high dimensions.

### 4. **Quasi-Newton (e.g., BFGS, L-BFGS)**

- **Use When**: Moderate to large-scale problems where full Hessian is unavailable.
- **Pros**: Efficient approximation of Hessian; widely used.
- **Cons**: Can still be expensive for very large parameter spaces.

### 5. **Conjugate Gradient**

- **Use When**: Large-scale problems where Hessian is impractical.
- **Pros**: Avoids matrix storage; good for sparse problems.
- **Cons**: Requires line search; can be unstable.

---

## 🔹 Second-Order and Derivative-Free Methods

These are typically more **robust**, though computationally more expensive. Often used when gradients are unavailable or unreliable.

### 6. **Expectation-Maximization (EM)**

- **Use When**: Latent variables or incomplete data are involved.
- **Pros**: Breaks complex likelihoods into simpler sub-problems.
- **Cons**: Slow convergence; can get stuck in local optima.

### 7. **Nelder-Mead (Simplex Method)**

- **Use When**: The likelihood is noisy, non-differentiable, or difficult to model analytically.
- **Pros**: No gradient required; simple to implement.
- **Cons**: Poor performance in high-dimensional spaces; local convergence only.

### 8. **Powell’s Method**

- **Use When**: Derivatives are not available but the function is smooth.
- **Pros**: Efficient in low dimensions.
- **Cons**: Slower than gradient-based methods.

---

## 🔹 Global Optimization Methods

For highly non-convex likelihood surfaces or when initialization is crucial.

### 9. **Simulated Annealing**

- **Use When**: Global optimization is needed; multiple local maxima.
- **Pros**: Escapes local optima.
- **Cons**: Slow; convergence is probabilistic.

### 10. **Genetic Algorithms (GA)**

- **Use When**: Discrete parameter spaces or rugged likelihood surfaces.
- **Pros**: Does not require differentiability.
- **Cons**: Expensive; sensitive to parameter tuning.

### 11. **Particle Swarm Optimization (PSO)**

- **Use When**: Parallelizability is desired and global search is critical.
- **Pros**: Easy to implement; global search.
- **Cons**: Often slower than gradient methods; convergence is not guaranteed.

---

## 🔹 Bayesian-Inspired and Hybrid Approaches

While not strictly MLE, these can approximate MLE solutions or improve robustness.

### 12. **Variational Inference (VI)**

- **Use When**: Posterior approximation is needed; high-dimensional latent models.
- **Pros**: Scalable; differentiable approximations to MLE/MAP.
- **Cons**: Requires approximation of the likelihood.

### 13. **Laplace Approximation**

- **Use When**: Approximate posterior near mode (MLE); used in Bayesian contexts.
- **Pros**: Fast for unimodal posteriors.
- **Cons**: Poor for multi-modal distributions.

## 🔹 Recommendations by Use Case

| Scenario                                           | Recommended Algorithm(s)        |
| -------------------------------------------------- | ------------------------------- |
| High-dimensional, differentiable model             | L-BFGS, Conjugate Gradient      |
| Large dataset (mini-batch required)                | SGD, Adam (SGD variant)         |
| Model with latent variables or missing data        | EM                              |
| Non-differentiable or noisy likelihood             | Nelder-Mead, Genetic Algorithms |
| Rugged likelihood landscape (many local optima)    | Simulated Annealing, PSO        |
| Probabilistic model, variational inference desired | VI, EM                          |
| Bayesian approximation near MLE                    | Laplace Approximation           |
