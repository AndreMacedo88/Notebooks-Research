# Import packages


```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Functions


```python
def L1(array1, array2):
    return round(sum(abs(a - b) for a, b in zip(array1, array2)), 4)


def L2(array1, array2):
    return round(math.sqrt(sum((a - b) ** 2 for a, b in zip(array1, array2))), 4)


def cosine(array1, array2):
    """Note that this is the cosine similarity and not the 'distance',
    as opposed to the previous notebook. distance = 1 - similarity

    """
    return round(
        np.dot(array1, array2)
        / (math.sqrt(sum(i**2 for i in array1)) * math.sqrt(sum(i**2 for i in array2))),
        4,
    )
```

# Create data


```python
cell1 = [0.4, 0.6]
cell1_larger = [0.8, 1.2]  # 2x larger
cell1_orthogonal = [-0.6, 0.4]  # 90 degree angle
```

```python
origin = np.array([0, 0])

plt.figure(figsize=(8, 6))

# Plot points
ax = plt.axes()

ax.scatter(
    [cell1[0]],
    [cell1[1]],
    s=100,
    alpha=1,
    zorder=3,
    label="cell1",
)
ax.scatter(
    [cell1_larger[0]],
    [cell1_larger[1]],
    s=100,
    alpha=1,
    zorder=3,
    label="cell1 2x larger",
)
ax.scatter(
    [cell1_orthogonal[0]],
    [cell1_orthogonal[1]],
    s=100,
    alpha=1,
    zorder=3,
    label="cell1 orthogonal",
)

# Draw vectors
ax.arrow(
    *origin,
    *cell1,
    linewidth=2,
    alpha=0.6,
)

ax.arrow(
    *origin,
    *cell1_larger,
    linewidth=2,
    alpha=0.6,
)

ax.arrow(
    *origin,
    *cell1_orthogonal,
    linewidth=2,
    alpha=0.6,
)

plt.title("Cell's gene expression distributions\n(connected to 0 for viewing angles)")
plt.legend()
plt.grid(True, alpha=0.3)
# plt.xlim(0, 1)
# plt.ylim(0, 1)
plt.show()
```

# Check differences in the results from the distance metrics


```python
print(f"L1 distance between cell1 and itself: {L1(cell1, cell1)}")
print(f"L2 distance between cell1 and itself: {L2(cell1, cell1)}")
print(f"Cosine similarity between cell1 and itself: {cosine(cell1, cell1)}")
```

```python
print(f"L1 distance between cell1 and cell1 2x larger: {L1(cell1, cell1_larger)}")
print(f"L2 distance between cell1 and cell1 2x larger: {L2(cell1, cell1_larger)}")
print(
    f"Cosine similarity between cell1 and cell1 2x larger: {cosine(cell1, cell1_larger)}"
)
```

```python
print(f"L1 distance between cell1 and cell1 orthogonal: {L1(cell1, cell1_orthogonal)}")
print(f"L2 distance between cell1 and cell1 orthogonal: {L2(cell1, cell1_orthogonal)}")
print(
    f"Cosine similarity between cell1 and cell1 orthogonal: {cosine(cell1, cell1_orthogonal)}"
)
```

```python
print(
    f"L1 distance between cell1 2x larger and cell1 orthogonal: {L1(cell1_larger, cell1_orthogonal)}"
)
print(
    f"L2 distance between cell1 2x larger and cell1 orthogonal: {L2(cell1_larger, cell1_orthogonal)}"
)
print(
    f"Cosine similarity between cell1 2x larger and cell1 orthogonal: {cosine(cell1_larger, cell1_orthogonal)}"
)
```
